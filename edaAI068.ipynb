```python
import numpy as np
np.random.seed(1)
arr=np.random.randint(10,size=(2,5))
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
print(arr.nbytes)
print(arr[1,4])
print(arr[-1,-2])
print(arr[:2,:3])
print(arr[1:2,:3])
print(arr[:,0])
print(arr[0,:])
```
